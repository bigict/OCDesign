```python
def identify_2d_front_ops(f1, f2, ops1 = lambda x,y:x>y, ops2 = lambda x,y:x>y):
    length = len(f1)
    front_indics = []
    for i in range(length):
        is_front = True
        for j in range(length):
            if i != j and ops1(f1[i],f1[j]) and ops2(f2[i],f2[j]):
                is_front = False
                    
        if is_front:
            front_indics.append(i)
    print(front_indics)
    return f1[front_indics], f2[front_indics]

def sorted_front(f1,f2,ops = '>', extend_t = ()):
    t_f12 = zip(f1,f2)
    print(t_f12)
    reverse = False
    if ops == '>': 
        reverse = True
    s_f12_sorted = sorted(t_f12, key=lambda x: x[0], reverse=reverse)
    
    return [i[0] for i in s_f12_sorted], [i[1] for i in s_f12_sorted]
```

```python
def identify_2d_front(f1, f2):
    length = len(f1)
    front_indics = []
    for i in range(length):
        is_front = True
        for j in range(length):
            if i != j and f1[j] >= f1[i] and f2[j] >= f2[i]:
                is_front = False
                    
        if is_front:
            front_indics.append(i)
    print(front_indics)
    return f1[front_indics], f2[front_indics]
def identify_pareto_front_3d(f1, f2, f3):
    front_indics = []
    for i in range(len(f1)):
        is_front = True
        for j in range(len(f1)):
            if i != j and f1[i] >= f1[j] and f2[i] >= f2[j] and f3[i] <= f3[j]:
                is_front = False
        if is_front:
            front_indics.append(i)
    return front_indics 
def fit_2d_front(f1, f2, deg = 2):
    med = np.median(f1)
    
    poly = np.polyfit(f1, f2, deg)
    parameter = poly
    p = np.poly1d(parameter)
    
    f1_fitted = np.arange(np.min(f1),np.max(f1),0.0001)
    f2_fitted = p(f1_fitted)

    # plt.plot(f1_fitted , f2_fitted, 'r--')
    # f1_fitted = np.concatenate([f1_fitted_1, f1_fitted_2], axis=0)
    # f2_fitted = np.concatenate([f2_fitted_1, f2_fitted_2], axis=0)
    
    return f1_fitted,f2_fitted
    
def plot_2d(f1, f2, select_indics, method, file = None, deg = 3, tight_layout = True, title = 'Pareto front of the Round 1', labels = None):
    # 创建基本的画布
    ax = plt.figure(figsize=figsize_cm)
    
    plt.xlabel(r'TM-score ($\!\!\uparrow\!\!$)')
    plt.ylabel(r'Solubility ($\!\!\uparrow\!\!$)')
    
    # 准备数据
    method1_indics_selected = [i for i in range(len(method)) if method[i] == "NeuralMRF" and select_indics[i] != "/"]
    method2_indics_selected = [i for i in range(len(method)) if method[i] != "NeuralMRF" and select_indics[i] != "/"]

    method1_indics_nselected = [i for i in range(len(method)) if method[i] == "NeuralMRF" and select_indics[i] == "/"]
    method2_indics_nselected = [i for i in range(len(method)) if method[i] != "NeuralMRF" and select_indics[i] == "/"]

    # 绘制step
    _front_f1, _front_f2 = identify_2d_front_ops(f1, f2, ops1 = lambda x,y:x<y, ops2 = lambda x,y:x<y)
    sorted_f1, sorted_f2 = sorted_front(_front_f1, _front_f2,">")
    plt.step(_front_f1, _front_f2,**step_args)

    if 'MRF' in title:plt.scatter(f1[method1_indics_nselected + method1_indics_selected], f2[method1_indics_nselected + method1_indics_selected], c='#6AAFF6',label='NeuralMRF',  marker = "^", **args)  
    elif 'LE' in title:plt.scatter(f1[method2_indics_nselected + method2_indics_selected], f2[method2_indics_nselected + method2_indics_selected], c='#F0C678',label='ProDESIGN-LE', marker = "o", **args)  
    else:raise ValueError("Neither MRF or LE result show!!!")
    plt.legend(loc = 'lower right')
    plt.scatter(f1[method1_indics_selected], f2[method1_indics_selected], label='NeuralMRF', marker = "^", **args1)
    plt.scatter(f1[method2_indics_selected], f2[method2_indics_selected], label='ProDESIGN-LE', marker = "o", **args1) 
    
    
    plt.gca().set_yticks(np.arange(0.48,1.0,0.08))
    plt.gca().set_xticks(np.arange(0.94, 0.98,0.006))
    ax.gca().yaxis.set_major_formatter(FormatStrFormatter('%.3f'))  # 试试使用固定小数位数
    ax.gca().xaxis.set_major_formatter(FormatStrFormatter('%.3f'))  # 试试使用固定小数位数
    ax.gca().invert_xaxis()
    ax.gca().invert_yaxis()
    plt.grid(**grad_args)
    plt.show()
    if file is not None:ax.savefig(file,dpi=300, bbox_inches='tight')
    return None 
```
