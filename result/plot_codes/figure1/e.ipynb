```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# config matplotlib parameters, and set all font size as 7pt
plt.rcParams.update({
    'font.size': 7,
    'axes.titlesize': 7,
    'axes.labelsize': 7,
    'xtick.labelsize': 7,
    'ytick.labelsize': 7,
    'legend.fontsize': 7,
    'figure.titlesize': 7
})

cm_to_inch = 1 / 2.54
figsize_cm = (7 * cm_to_inch, 4.25 * cm_to_inch)  # (9cm, 9cm)
fig_size = figsize_cm
fig_args = {"figsize":fig_size}
tick_args = {"pad":0}
lable_args = {"labelpad" : -5}
save_args = {"bbox_inches":'tight', "dpi":300}
```

```python
# read the CSV file
csv_file = '../data/successtate.csv'
df = pd.read_csv(csv_file)
print(df)
```

```python
# config color set
colors = ["#A0E399","#B0A6D8","#FF3333","#F09890"] 

# plot success rate bar
plt.figure(**fig_args)
sns.barplot(data=df, x='Round', y='Success rate', hue='Property', palette = colors)

# config x&y label, for we add the lable in ppt
plt.ylim(0, 100)
plt.xlabel(None)
plt.ylabel(None)
sns.despine(top=True, right=True)

# show legend
plt.grid(False)
plt.legend(bbox_to_anchor=(1, 1.05), loc='upper left',frameon=False, handlelength=1.5)

plt.gca().tick_params(axis='x', labelbottom=False)

# save figure
plt.tight_layout()
plt.savefig('success_rate.svg', dpi=300)
plt.show()
```
