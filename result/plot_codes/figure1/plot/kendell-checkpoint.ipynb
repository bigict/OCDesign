```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# ── Load data ──────────────────────────────────────────────────────────────
df = pd.read_csv("round2-3_pareto.csv", skipinitialspace=True)
df.columns = df.columns.str.strip()
df["IgG-Binding"] = df["IgG-Binding"].str.strip()

# Extract round from Name: PDN<round><id>, e.g. PDN11 → round 1
df["Round"] = df["Name"].str.extract(r"PDN(\d)\d").astype(int)
# Extract within-round id from Name: PDN<round><id>, e.g. PDN11 → id 1
df["RoundID"] = df["Name"].str.extract(r"PDN\d(\d)").astype(int)

df["Round"] = [2] * 14 + [3] * 14
df["RoundID"] = [2] * 14 + [3] * 14

df = df[~df['Name'].str.contains('PDN35', na=False)]
# ── Pareto helpers ─────────────────────────────────────────────────────────
def pareto_front(costs):
    """
    costs: (N, 2) array where LOWER is better for both dims.
    Returns boolean mask of Pareto-optimal points.
    """
    is_pareto = np.ones(len(costs), dtype=bool)
    for i, c in enumerate(costs):
        if is_pareto[i]:
            # dominated if another point is <= in all dims and < in at least one
            dominated = np.all(costs >= c, axis=1) & np.any(costs > c, axis=1)
            dominated[i] = False
            is_pareto[dominated] = False
    return is_pareto


def curvature_from_points(pts):
    """
    Approximate signed curvature at each interior point of a polyline.
    pts: (N, 2) sorted array.
    Returns arrays x, y, kappa (length N-2).
    """
    if len(pts) < 3:
        return pts[:, 0], pts[:, 1], np.array([])
    x, y = pts[:, 0], pts[:, 1]
    kappas = []
    for i in range(1, len(pts) - 1):
        x1, y1 = x[i-1], y[i-1]
        x2, y2 = x[i],   y[i]
        x3, y3 = x[i+1], y[i+1]
        # Menger curvature
        a = np.hypot(x2-x1, y2-y1)
        b = np.hypot(x3-x2, y3-y2)
        c = np.hypot(x3-x1, y3-y1)
        area = abs((x2-x1)*(y3-y1) - (x3-x1)*(y2-y1)) / 2
        denom = a * b * c
        kappas.append(2 * area / denom if denom > 1e-12 else 0.0)
    return x[1:-1], y[1:-1], np.array(kappas)


def minmax_normalize_points(pts, ref_pts=None):
    if len(pts) == 0:
        return pts.copy()
    if ref_pts is None:
        ref_pts = pts
    mins = np.min(ref_pts, axis=0)
    maxs = np.max(ref_pts, axis=0)
    spans = maxs - mins
    spans[spans == 0] = 1.0
    return (pts - mins) / spans


def conflict_from_kendall(a, b):
    """
    Conflict score in [0, 1] from Kendall rank correlation:
      C = (1 - tau) / 2
    C close to 1: strong conflict; close to 0: strong synergy.
    """
    valid = pd.DataFrame({"a": a, "b": b}).dropna()
    if len(valid) < 2:
        return np.nan, np.nan
    tau = valid["a"].corr(valid["b"], method="kendall")
    if pd.isna(tau):
        return np.nan, np.nan
    return tau, (1 - tau) / 2


def safe_legend(ax, **kwargs):
    handles, labels = ax.get_legend_handles_labels()
    if handles:
        ax.legend(**kwargs)

def fmt_metric(value, digits=3):
    if pd.isna(value):
        return "-"
    return f"{value:.{digits}f}"


# ── Axis pairs (x, y) — both transformed so LOWER = better ────────────────
# RMSD_VAR: lower better  → use as-is
# MMGBS:    lower better  → use as-is  (values are negative, so "lower" means more negative)
# Solubility: higher better → negate so lower = better
df["neg_Sol"] = -df["Solubitlity"]

pairs = [
    ("RMSD_VAR", "MMGBS",    "RMSD_VAR",   "MMGBS"),
    ("RMSD_VAR", "neg_Sol",  "RMSD_VAR",   "Solubility (negated)"),
    ("MMGBS",    "neg_Sol",  "MMGBS",       "Solubility (negated)"),
]

rounds   = sorted(df["Round"].unique())
colors   = {1: "#1f77b4", 2: "#ff7f0e", 3: "#2ca02c"}
markers  = {1: "o",       2: "s",       3: "^"}

# ── Plot ───────────────────────────────────────────────────────────────────
fig, axes = plt.subplots(
    2, len(pairs),
    figsize=(6 * len(pairs), 11),
    gridspec_kw={"height_ratios": [3, 1]},
)

for col, (xcol, ycol, xlabel, ylabel) in enumerate(pairs):
    ax_scatter  = axes[0, col]
    ax_curv_abs = axes[1, col]
    tau_all, conflict_all = conflict_from_kendall(df[xcol], df[ycol])

    for rnd in rounds:
        sub = df[df["Round"] == rnd][["Name", "RoundID", xcol, ycol]].dropna()
        if sub.empty:
            continue
        tau_rnd, conflict_rnd = conflict_from_kendall(sub[xcol], sub[ycol])
        round_label = (
            f"Round {rnd} (C={conflict_rnd:.2f})"
            if not pd.isna(conflict_rnd)
            else f"Round {rnd}"
        )
        pts_all = sub[[xcol, ycol]].values

        # scatter all points
        ax_scatter.scatter(
            pts_all[:, 0], pts_all[:, 1],
            c=colors[rnd], marker=markers[rnd],
            s=130, alpha=0.82, label=round_label,
        )
        for _, row in sub.iterrows():
            ax_scatter.annotate(
                f"{int(row['RoundID'])}",
                (row[xcol], row[ycol]),
                textcoords="offset points",
                xytext=(0, 0),
                ha="center",
                va="center",
                fontsize=8,
                fontweight="bold",
                color="white",
                alpha=1.0,
                zorder=11,
            )

        # Pareto front
        mask = pareto_front(pts_all)
        front_df = sub.loc[mask].copy()
        front_df = front_df.sort_values(xcol)  # sort by x for line plot
        front = front_df[[xcol, ycol]].values

        ax_scatter.plot(
            front[:, 0], front[:, 1],
            color=colors[rnd], linewidth=1.8, linestyle="--",
        )
        ax_scatter.scatter(
            front[:, 0], front[:, 1],
            c=colors[rnd], marker=markers[rnd],
            s=210, edgecolors="black", linewidths=1.0, zorder=6,
        )

        # curvature
        front_norm = minmax_normalize_points(front, ref_pts=pts_all)
        xk, yk, kappa = curvature_from_points(front_norm)
        if len(kappa):
            ax_curv_abs.plot(xk, kappa, color=colors[rnd],
                             marker=markers[rnd], markersize=9,
                             linewidth=1.5, label=round_label)
            curv_ids = front_df["RoundID"].to_numpy()[1:-1]
            for xi, ki, rid in zip(xk, kappa, curv_ids):
                ax_curv_abs.annotate(
                    f"{int(rid)}",
                    (xi, ki),
                    textcoords="offset points",
                    xytext=(0, 0),
                    ha="center",
                    va="center",
                    fontsize=7.5,
                    fontweight="bold",
                    color="white",
                    alpha=1.0,
                    zorder=12,
                )

    # labels
    ax_scatter.set_xlabel(xlabel, fontsize=11)
    ax_scatter.set_ylabel(ylabel, fontsize=11)
    title = f"Pareto front: {xlabel} vs {ylabel}"
    if not pd.isna(conflict_all):
        title += f"\nConflict(all)={conflict_all:.2f}"
    ax_scatter.set_title(title, fontsize=12)
    safe_legend(ax_scatter, fontsize=9)
    ax_scatter.grid(True, alpha=0.3)

    ax_curv_abs.set_xlabel(xlabel, fontsize=10)
    ax_curv_abs.set_ylabel("Curvature κ", fontsize=10)
    ax_curv_abs.set_title("Pareto front curvature (normalized axes)", fontsize=10)
    safe_legend(ax_curv_abs, fontsize=8)
    ax_curv_abs.grid(True, alpha=0.3)

fig.suptitle(
    "Per-round Pareto fronts & curvature\n"
    "(RMSD_VAR ↓  MMGBS ↓  Solubility ↑)",
    fontsize=14, y=1.01,
)
plt.tight_layout()
plt.savefig("pareto_analysis.png", dpi=150, bbox_inches="tight")
print("Saved → pareto_analysis.png")


# ── Plot (all rounds combined for Pareto front and curvature) ──────────────
fig2, axes2 = plt.subplots(
    2, len(pairs),
    figsize=(6 * len(pairs), 11),
    gridspec_kw={"height_ratios": [3, 1]},
)

for col, (xcol, ycol, xlabel, ylabel) in enumerate(pairs):
    ax_scatter = axes2[0, col]
    ax_curv_abs = axes2[1, col]
    tau_all, conflict_all = conflict_from_kendall(df[xcol], df[ycol])

    # Plot all points by round (keep round identity)
    for rnd in rounds:
        sub = df[df["Round"] == rnd][["Name", "RoundID", xcol, ycol]].dropna()
        if sub.empty:
            continue
        ax_scatter.scatter(
            sub[xcol], sub[ycol],
            c=colors[rnd], marker=markers[rnd],
            s=130, alpha=0.82, label=f"Round {rnd}",
        )
        for _, row in sub.iterrows():
            ax_scatter.annotate(
                f"{int(row['RoundID'])}",
                (row[xcol], row[ycol]),
                textcoords="offset points",
                xytext=(0, 0),
                ha="center",
                va="center",
                fontsize=8,
                fontweight="bold",
                color="white",
                alpha=1.0,
                zorder=11,
            )

    # Pareto front from ALL rounds combined
    all_sub = df[["Name", "Round", "RoundID", xcol, ycol]].dropna().copy()
    all_pts = all_sub[[xcol, ycol]].values
    mask_all = pareto_front(all_pts)
    front_all_df = all_sub.loc[mask_all].copy().sort_values(xcol)
    front_all = front_all_df[[xcol, ycol]].values

    ax_scatter.plot(
        front_all[:, 0], front_all[:, 1],
        color="black", linewidth=2.0, linestyle="--",
        label="All-round Pareto front",
    )
    for rnd in rounds:
        front_rnd = front_all_df[front_all_df["Round"] == rnd]
        if front_rnd.empty:
            continue
        ax_scatter.scatter(
            front_rnd[xcol], front_rnd[ycol],
            c=colors[rnd], marker=markers[rnd],
            s=220, edgecolors="black", linewidths=1.0, zorder=7,
        )

    # Curvature from ALL-round Pareto front
    front_all_norm = minmax_normalize_points(front_all, ref_pts=all_pts)
    xk, yk, kappa = curvature_from_points(front_all_norm)
    if len(kappa):
        ax_curv_abs.plot(
            xk, kappa,
            color="black", linestyle="--", linewidth=1.7,
            label="All-round Pareto front",
        )
        curv_df = front_all_df.iloc[1:-1].copy()
        curv_df["xk"] = xk
        curv_df["kappa"] = kappa
        for rnd in rounds:
            curv_rnd = curv_df[curv_df["Round"] == rnd]
            if curv_rnd.empty:
                continue
            ax_curv_abs.scatter(
                curv_rnd["xk"], curv_rnd["kappa"],
                c=colors[rnd], marker=markers[rnd],
                s=90, edgecolors="black", linewidths=0.9,
                label=f"Round {rnd}",
            )
            for _, row in curv_rnd.iterrows():
                ax_curv_abs.annotate(
                    f"{int(row['RoundID'])}",
                    (row["xk"], row["kappa"]),
                    textcoords="offset points",
                    xytext=(0, 0),
                    ha="center",
                    va="center",
                    fontsize=7.5,
                    fontweight="bold",
                    color="white",
                    alpha=1.0,
                    zorder=12,
                )

    # labels
    ax_scatter.set_xlabel(xlabel, fontsize=11)
    ax_scatter.set_ylabel(ylabel, fontsize=11)
    title = f"All-round Pareto front: {xlabel} vs {ylabel}"
    if not pd.isna(conflict_all):
        title += f"\nConflict(all)={conflict_all:.2f}"
    ax_scatter.set_title(title, fontsize=12)
    safe_legend(ax_scatter, fontsize=9)
    ax_scatter.grid(True, alpha=0.3)

    ax_curv_abs.set_xlabel(xlabel, fontsize=10)
    ax_curv_abs.set_ylabel("Curvature κ", fontsize=10)
    ax_curv_abs.set_title("All-round Pareto front curvature (normalized axes)", fontsize=10)
    safe_legend(ax_curv_abs, fontsize=8)
    ax_curv_abs.grid(True, alpha=0.3)

fig2.suptitle(
    "All-round Pareto fronts & curvature\n"
    "(points still indicate source round)",
    fontsize=14, y=1.01,
)
plt.tight_layout()
plt.savefig("pareto_analysis_allround.png", dpi=150, bbox_inches="tight")
print("Saved → pareto_analysis_allround.png")


# ── Metrics table PNG ───────────────────────────────────────────────────────
conflict_rows = []

for xcol, ycol, xlabel, ylabel in pairs:
    pair_label = f"{xlabel} vs {ylabel}"
    tau_all, conflict_all = conflict_from_kendall(df[xcol], df[ycol])

    conflict_row = {
        "Pair": pair_label,
        "All": fmt_metric(conflict_all, 2),
    }

    for rnd in rounds:
        sub = df[df["Round"] == rnd].copy()
        _, conflict_rnd = conflict_from_kendall(sub[xcol], sub[ycol])
        conflict_row[f"R{rnd}"] = fmt_metric(conflict_rnd, 2)

    conflict_rows.append(conflict_row)

conflict_df = pd.DataFrame(conflict_rows)

fig3, axes3 = plt.subplots(
    2, 1,
    figsize=(14, 6.5),
    gridspec_kw={"height_ratios": [3.2, 1]},
)
axes3[0].axis("off")
axes3[1].axis("off")

conflict_table = axes3[0].table(
    cellText=conflict_df.values,
    colLabels=conflict_df.columns,
    cellLoc="center",
    loc="center",
)
conflict_table.auto_set_font_size(False)
conflict_table.set_fontsize(11)
conflict_table.scale(1, 1.8)

for (row, col), cell in conflict_table.get_celld().items():
    if row == 0:
        cell.set_facecolor("#d9e6f2")
        cell.set_text_props(weight="bold")
    elif col == 0:
        cell.set_facecolor("#f4f4f4")

axes3[0].set_title(
    "Conflict Summary Table\nC = (1 - Kendall tau) / 2",
    fontsize=13,
    pad=12,
)
axes3[1].text(
    0.01, 0.92,
    "How to read C:\n"
    "C close to 1.0: stronger conflict between the two metrics\n"
    "C around 0.5: weak or no clear monotonic relationship\n"
    "C close to 0.0: stronger synergy / lower conflict",
    fontsize=12,
    va="top",
)

fig3.suptitle("Metrics Summary", fontsize=15, y=0.98)
plt.tight_layout()
plt.savefig("pareto_metrics_table.png", dpi=180, bbox_inches="tight")
print("Saved → pareto_metrics_table.png")
```
