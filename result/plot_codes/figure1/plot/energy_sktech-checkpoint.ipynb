```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import cm
from mpl_toolkits.mplot3d import Axes3D

plt.rcParams.update({
    'font.size': 7,
    'axes.titlesize': 7,
    'axes.labelsize': 7,
    'xtick.labelsize': 7,
    'ytick.labelsize': 7,
    'legend.fontsize': 7,
    'figure.titlesize': 7
})

cm_to_inch = 1 / 2.54
figsize_cm = (9 * cm_to_inch, 9 * cm_to_inch)  # (9cm, 9cm)
fig_size = figsize_cm
# 设置画布
fig = plt.figure(figsize=fig_size)
ax = fig.add_subplot(111, projection='3d')

# 网格密度
x = np.linspace(0, 1, 200)
y = np.linspace(0, 1, 200)
X, Y = np.meshgrid(x, y)

# 高斯峰函数
def gaussian_peak(x, y, x0, y0, amp, sigma):
    return amp * np.exp(-((x - x0)**2 + (y - y0)**2) / (2 * sigma**2))

# 三个峰的新中心位置（彼此更靠近，减少平坦区域）
peak_a = gaussian_peak(X, Y, 0.28, 0.28, 2.0, 0.05)
peak_b = gaussian_peak(X, Y, 0.52, 0.72, 1.5, 0.15)
peak_c = gaussian_peak(X, Y, 0.76, 0.34, 1.0, 0.08)

# 合并光滑地形
Z_smooth = peak_a + peak_b + peak_c

# 增加随机抖动（幅度提高至 0.04，让平坦区域更不平坦）
np.random.seed(42)
noise = np.random.normal(0, 0.02, Z_smooth.shape)
Z = Z_smooth + noise

# 绘制曲面
surf = ax.plot_surface(X, Y, Z, cmap=cm.coolwarm, linewidth=0, antialiased=True, alpha=0.9)

# 颜色条（可选，若不想要可注释）
# fig.colorbar(surf, ax=ax, shrink=0.5, aspect=20, label='Elevation')

# 标记峰顶位置（根据新中心）
peaks = {
    'a': (0.28, 0.28),
    'b': (0.52, 0.72),
    'c': (0.76, 0.34)
}
# for label, (px, py) in peaks.items():
#     pz_smooth = gaussian_peak(px, py, 0.28, 0.28, 1.0, 0.08) + \
#                 gaussian_peak(px, py, 0.52, 0.72, 1.0, 0.08) + \
#                 gaussian_peak(px, py, 0.76, 0.34, 1.0, 0.08)
#     ix = int(px * (x.size - 1))
#     iy = int(py * (y.size - 1))
#     pz_noise = noise[iy, ix]
#     pz = pz_smooth + pz_noise
#     ax.scatter(px, py, pz, color='red', s=50, edgecolor='k', zorder=10)
#     ax.text(px, py, pz + 0.05, label, fontsize=14, fontweight='bold',
#             ha='center', va='bottom', color='black')

# 彻底隐藏坐标轴、框线、刻度等
ax.set_axis_off()

# 调整视角
ax.view_init(elev=25, azim=-60)

plt.tight_layout()
plt.savefig('energy_sketch.svg', dpi= 300)
plt.show()
```
