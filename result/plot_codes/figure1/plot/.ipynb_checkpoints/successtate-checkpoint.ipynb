```python
csv_file = 'successtate.csv'
df = pd.read_csv(csv_file)
# 确保 Round 列按顺序排序（Round0, Round1, Round2, Round3）
df
```

```python
round_order = [f"Round{i}" for i in range(4)]
df['Round'] = pd.Categorical(df['Round'], categories=round_order, ordered=True)

# 设置绘图风格
sns.set_theme(style="whitegrid")

# 绘制分组柱状图
plt.figure(figsize=(8, 5))
sns.barplot(data=df, x='Round', y='Success rate', hue='Property', palette='viridis')

# 添加标题和标签
plt.ylabel('Success rate (%)', fontsize=12)
# 显示图例
plt.legend(title='Property', bbox_to_anchor=(1.05, 1), loc='upper left')

# 调整布局，避免图例被裁剪
plt.tight_layout()

# 显示图形
plt.show()
```
