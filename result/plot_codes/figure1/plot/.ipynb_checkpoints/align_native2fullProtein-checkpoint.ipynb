```python
MACARON_GRAY = "0xB0B0B0"

def hex_to_rgb(hex_str):
    hex_val = int(hex_str.lstrip('#'), 16)
    return [((hex_val >> 16) & 0xFF) / 255.0, ((hex_val >> 8) & 0xFF) / 255.0, (hex_val & 0xFF) / 255.0]

def interpolate_color(start_hex, end_hex, fraction):
    c1 = hex_to_rgb(start_hex)
    c2 = hex_to_rgb(end_hex)
    return [c1[i] + (c2[i] - c1[i]) * fraction for i in range(3)]

def get_vintage_gradient():
    return [
        ("0xF09890", "0xF6B5B0"),
        ("0xFBE579", "0xFFF8B5"),
        ("0xA0E399", "0xB2E8AC"),
        ("0xB0A6D8", "0xBFB6E0"),
    ]
def setup_pub_style():
    cmd.bg_color("white")
    cmd.set("orthoscopic", 1)
    cmd.set("ray_trace_mode", 1)
    cmd.set("ray_trace_color", "black")
    cmd.set("ambient", 1)
    cmd.set("specular", 0)
    cmd.set("ray_shadows", 0)
    cmd.set("cartoon_fancy_helices", 1)
    cmd.viewport(1000, 1000)
      
def load_pdbs(native_pdb, full_protein):
    for i in range(1, 4 + 1):cmd.load(native, 'native' + f'{i}')
    cmd.load(full_protein, 'fullprotein')
```
