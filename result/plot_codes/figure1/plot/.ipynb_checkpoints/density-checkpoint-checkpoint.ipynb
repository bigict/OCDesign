```python
coords = reduce_symmetric_matrix(matrix, seed = 742, labels = labels,perplexity = 4)
# 你的数据（29个点）
x = coords[:,0]
y = coords[:,1]
def get_density(x,y):
    xy = np.vstack([x, y])
    kde = gaussian_kde(xy)
    xi = np.linspace(x.min(), x.max(), 100)
    yi = np.linspace(y.min(), y.max(), 100)
    XI, YI = np.meshgrid(xi, yi)
    positions = np.vstack([XI.ravel(), YI.ravel()])
    density = kde(positions).reshape(XI.shape)
    return XI, YI, density
def plot_density(x, y, title, color):
    print(f'========================================================')
    plt.figure(figsize=(6,5))
    XI, YI, density = get_density(x, y)
    plt.contourf(XI, YI, density, levels=20, cmap='plasma')
    plt.colorbar(label='Density')
    plt.scatter(x, y, c='black', s=20, alpha=0.6)
    plt.xlabel('X'); plt.ylabel('Y')
    plt.title('KDE Density Plot')
    plt.show()
    
```

```python
for i in range(1,46):
    coords = reduce_symmetric_matrix(matrix, seed = 742, labels = labels,perplexity = i)
    # 你的数据（29个点）
    x = coords[:,0]
    y = coords[:,1]
    def get_density(x,y):
        xy = np.vstack([x, y])
        kde = gaussian_kde(xy)
        xi = np.linspace(x.min(), x.max(), 100)
        yi = np.linspace(y.min(), y.max(), 100)
        XI, YI = np.meshgrid(xi, yi)
        positions = np.vstack([XI.ravel(), YI.ravel()])
        density = kde(positions).reshape(XI.shape)
        return XI, YI, density
    def plot_density(x, y, title, color):
        print(f'========================================================')
        plt.figure(figsize=(6,5))
        XI, YI, density = get_density(x, y)
        plt.contourf(XI, YI, density, levels=20, cmap='plasma')
        plt.colorbar(label='Density')
        plt.scatter(x, y, c='black', s=20, alpha=0.6)
        plt.xlabel('X'); plt.ylabel('Y')
        plt.title('KDE Density Plot')
        plt.show()
        
    # plot all data on a plot 
    zips = [
        (0, -1, 'all', '#B6B3D6'),
        # (0,0 + 1,'native', '#D5D3DE'),
        (1,  1  + 8, 'round0', '#D6B3D6'),
        (9,  9  + 10, 'round1', '#B6B3D6'),
        (19, 19 + 14,'round2', '#F6DFD6'),
        (33, 33 + 14, 'round3', '#E9687A')]
        
    np.random.seed(42)
    data = coords
    data0 = coords[zips[1][0]:zips[1][1]]
    data1 = coords[zips[2][0]:zips[2][1]]
    data2 = coords[zips[3][0]:zips[3][1]]
    data3 = coords[zips[4][0]:zips[4][1]]
    # ------------------------------
    
    # 统一坐标范围（所有数据的最小/最大 x, y）
    all_x = coords[:,0]
    all_y = coords[:,1]
    x_min, x_max = all_x.min(), all_x.max()
    y_min, y_max = all_y.min(), all_y.max()
    
    # interpolate the values while the designs only 38, and a 
    grid_size = 30
    xi = np.linspace(x_min, x_max, grid_size)
    yi = np.linspace(y_min, y_max, grid_size)
    XI, YI = np.meshgrid(xi, yi)
    positions = np.vstack([XI.ravel(), YI.ravel()])
    
    # 定义绘图函数：计算单个数据集的KDE并返回网格密度
    def get_kde_grid(data, positions):
        kde = gaussian_kde(data.T)          # data.T 形状 (2, N)
        density = kde(positions).reshape(XI.shape)
        return density
    
    # 计算三个数据集的密度
    density0 = get_kde_grid(data0, positions)
    density1 = get_kde_grid(data1, positions)
    density2 = get_kde_grid(data2, positions)
    density3 = get_kde_grid(data3, positions)
    
    # ------------------------------
    # 绘图
    colors = ['blue', 'green',  'Orange','red']
    fig = plt.figure(figsize=fig_size)
    levels = 3
    fill = False
    if fill:
        # fill lines
        cmaps = ['Blues', 'Greens','Oranges','Reds']
        plt.contourf(XI, YI, density0, levels=1, alpha=0.3, cmap=cmaps[0])
        plt.contourf(XI, YI, density1, levels=2, alpha=0.3, cmap=cmaps[1])
        plt.contourf(XI, YI, density2, levels=3, alpha=0.3, cmap=cmaps[2])
        plt.contourf(XI, YI, density3, levels=3, alpha=0.3, cmap=cmaps[3])
        
    # plot lines
    contour0 = plt.contour(XI, YI, density0, levels=0, colors=colors[0], linewidths=1.5, alpha=0.8)
    contour1 = plt.contour(XI, YI, density1, levels=2, colors=colors[1], linewidths=1.5, alpha=0.8)
    contour2 = plt.contour(XI, YI, density2, levels=3, colors=colors[2], linewidths=1.5, alpha=0.8)
    contour3 = plt.contour(XI, YI, density3, levels=2, colors=colors[3], linewidths=1.5, alpha=0.8)
    
    # scatter the designs 
    plt.scatter(data0[:,0], data0[:,1], c=colors[0], s=10, alpha=0.9, label='Round 0')
    plt.scatter(data1[:,0], data1[:,1], c=colors[1], s=10, alpha=0.9, label='Round 1')
    plt.scatter(data2[:,0], data2[:,1], c=colors[2], s=10, alpha=0.9, label='Round 2')
    plt.scatter(data3[:,0], data3[:,1], c=colors[3], s=10, alpha=0.9, label='Round 3')
    
    plt.xlim(None)
    plt.ylim(None)
    
    # plt.legend()
    plt.xticks([])
    plt.yticks([])
    plt.title(f'Protein space {i}')
    plt.axis('off')
    # plt.savefig('equal_line.svg', dpi = 300)
    plt.show()
```

```python
# plot all data on a plot 
zips = [
    (0, -1, 'all', '#B6B3D6'),
    # (0,0 + 1,'native', '#D5D3DE'),
    (1,10 + 1, 'round1', '#B6B3D6'),
    (11, 11 + 14,'round2', '#F6DFD6'),
    (25, 25 + 14, 'round3', '#E9687A')]
    
np.random.seed(42)
data0 = coords
data1 = coords[zips[1][0]:zips[1][1]]
data2 = coords[zips[2][0]:zips[2][1]]
data3 = coords[zips[3][0]:zips[3][1]]
# ------------------------------

# 统一坐标范围（所有数据的最小/最大 x, y）
all_x = coords[:,0]
all_y = coords[:,1]
x_min, x_max = all_x.min(), all_x.max()
y_min, y_max = all_y.min(), all_y.max()

# interpolate the values while the designs only 38, and a 
grid_size = 30
xi = np.linspace(x_min, x_max, grid_size)
yi = np.linspace(y_min, y_max, grid_size)
XI, YI = np.meshgrid(xi, yi)
positions = np.vstack([XI.ravel(), YI.ravel()])

# 定义绘图函数：计算单个数据集的KDE并返回网格密度
def get_kde_grid(data, positions):
    kde = gaussian_kde(data.T)          # data.T 形状 (2, N)
    density = kde(positions).reshape(XI.shape)
    return density

# 计算三个数据集的密度
density0 = get_kde_grid(data0, positions)

# ------------------------------
# 绘图
colors = ['blue', 'green', 'red']
fig = plt.figure(figsize=fig_size)

# plot lines
contour0 = plt.contour(XI, YI, density0, levels=2, colors='gray', linewidths=1.5, alpha=0.8)

# scatter the designs 
plt.scatter(data1[:,0], data1[:,1], c=colors[0], s=10, alpha=0.9, label='Round 1')
plt.scatter(data2[:,0], data2[:,1], c=colors[1], s=10, alpha=0.9, label='Round 2')
plt.scatter(data3[:,0], data3[:,1], c=colors[2], s=10, alpha=0.9, label='Round 3')

plt.xlim(None)
plt.ylim(None)

# plt.legend()
plt.xticks([])
plt.yticks([])
plt.title('Density')
plt.axis('off')
plt.savefig('equal_line.svg', dpi = 300)
plt.show()

# # ------------------------------
# 方法B：使用半透明填充等高线（如果确实需要填充）
# 注意：填充会互相遮挡，建议设置较低 alpha 值，且只保留一个填充或使用线图为主
plt.figure(figsize=fig_size)
plt.contourf(XI, YI, density0, levels=2, alpha=0.3, cmap='gray')
# 添加线框提高可读性
plt.contour(XI, YI, density0, levels=2, colors=colors[0], linewidths=0.8)

plt.scatter(data1[:,0], data1[:,1], c=colors[0], s=10, alpha=0.9, label='Round 1')
plt.scatter(data2[:,0], data2[:,1], c=colors[1], s=10, alpha=0.9, label='Round 2')
plt.scatter(data3[:,0], data3[:,1], c=colors[2], s=10, alpha=0.9, label='Round 3')

# plt.legend()
plt.xticks([])
plt.yticks([])
plt.title('Density')
plt.axis('off')
# plt.colorbar()
plt.savefig('filled_equal_line.svg', dpi = 300)
plt.show()
```
