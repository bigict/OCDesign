```python
import sys
import json
def parser_seq(seq_file):
    with open(seq_file) as f:
        lines = [line.strip() for line in f.readlines()]
    for line in lines:names = [line[1:] for line in lines if line.startswith('>')]
    for line in lines:seqs  = [line for line in lines if not line.startswith('>')]

    seq_dict = {}
    for n,s in zip(names, seqs):seq_dict.update({n:s})
    return seq_dict
def cat_seq(seq_dict, info_dict, json_file = 'r3.json'):
    output_json = []
    for n,s in seq_dict.items():
        seq_w_linker = info_dict['linker_head'] + s + info_dict['linker_tail']
        seq_wo_linker = s
        print(f"{n}_with_linker")
        print(seq_w_linker * 4 + ':' + info_dict['igg_seq'])
        print(f"{n}_without_linker")
        print(seq_wo_linker * 4+ ':' + info_dict['igg_seq'])
        with_linker = {
                  "name": f"{n}_with_linker-igg",
                  "modelSeeds": [],
                  "sequences": [
                        {
                        "proteinChain": {
                        "sequence":seq_w_linker * 4,
                        "count": 1
                                }
                        },
                        {
                        "proteinChain": {
                        "sequence":info_dict['igg_seq'],
                        "count": 1
                                }
                        }
                        ],
                  "dialect": "alphafoldserver",
                  "version": 1
                }
        without_linker = {
                  "name": f"{n}_without_linker-igg",
                  "modelSeeds": [],
                  "sequences": [
                        {
                        "proteinChain": {
                        "sequence":seq_wo_linker * 4,
                        "count": 1
                                }
                        },
                        {
                        "proteinChain": {
                        "sequence":info_dict['igg_seq'],
                        "count": 1
                                }
                        }
                        ],
                  "dialect": "alphafoldserver",
                  "version": 1
                }
        
        output_json.append(with_linker)
        output_json.append(without_linker)
        with open(json_file, 'w') as f:
            json.dump(output_json, f)
```
