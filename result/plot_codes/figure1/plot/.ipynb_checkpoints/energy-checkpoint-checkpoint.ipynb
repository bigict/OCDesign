```python
'''
load data include design fastas, lables, and computational metric
PDN35: a isolate data that with the RMSD varaince > 10 while others lower than 3
'''
def reduce_matrix(matrix, seed = 42, labels = None, perplexity = None):
    import numpy as np
    import matplotlib.pyplot as plt
    from sklearn.manifold import MDS
    from sklearn.manifold import TSNE
    
    # MDS 降维到 2 维
    mds = MDS(n_components=2, random_state=seed, dissimilarity='precomputed')
    tsne = TSNE(n_components=2, random_state=seed, perplexity=perplexity)
    pca = PCA(n_components=2)
    
    scaler = StandardScaler()   # 形状仍然是 (46, 46)
    
    coords = mds.fit_transform(matrix)   # 形状 (29, 2)
    # coords = tsne.fit_transform(matrix)
    # coords = pca.fit_transform(scaler.fit_transform(matrix)) 
    # theta = np.radians(135)
    # rot_matrix = np.array([[np.cos(theta), -np.sin(theta)],
    #                        [np.sin(theta),  np.cos(theta)]])
    
    # new_coords =  coords @ rot_matrix
    # print(new_coords.shape)
    return coords
    
def preprocess(csv_file, perplexity = 30):
    source_df = pd.read_csv(csv_file)
    
    with35 = True
    if not with35:df = source_df[~source_df['Name'].str.contains('PDN35', na=False)]
    else:df = source_df
        
    labels = df["Name"]
    fasta_list = list(df['Sequence'])
   
    matrix = 1 - get_identity_matrix(fasta_list)
    # coords = reduce_symmetric_matrix(matrix, seed = 742, labels = labels,perplexity = 10)
    coords = reduce_matrix(matrix, seed = 742, labels = labels,perplexity = perplexity)
    return coords, source_df
```

```python
'''
plot the 3d energe landscape
    x: reduced coords 0
    y: reduced coords 0 
    z: rmsd var
    energy:mmgbsa

    difficulty: too litle points (29), interpolation is needed
'''
# load the rmsd values
# load the energy values
# interpolate the energy value
def load_sequences_list(fasta_file):
    with open(fasta_file) as f:
        sequences_list = [line.strip() for line in f.readlines()]
    return sequences_list
def minmax_scale(data, feature_range=(0, 1)):
    """
    将数据缩放到指定区间，默认 [0, 1]。
    参数：
        data : 一维或二维 numpy 数组，若为二维，每列是一个特征
        feature_range : 元组 (min, max)，目标区间
    返回：
        data_scaled : 缩放后的数据
        params : 字典，包含每个特征的最小值和最大值，用于逆变换
    """
    data = np.asarray(data)
    if data.ndim == 1:
        data = data.reshape(-1, 1)
        single_feature = True
    else:
        single_feature = False

    min_vals = data.min(axis=0)
    max_vals = data.max(axis=0)
    range_vals = max_vals - min_vals
    # 防止除以零（所有值相同的情况）
    range_vals[range_vals == 0] = 1.0

    target_min, target_max = feature_range
    data_scaled = (data - min_vals) / range_vals  # 先缩放到 [0,1]
    data_scaled = data_scaled * (target_max - target_min) + target_min

    params = {
        'min': min_vals,
        'max': max_vals,
        'range': range_vals,
        'target_min': target_min,
        'target_max': target_max
    }
    if single_feature:
        data_scaled = data_scaled.flatten()
    return data_scaled, params

def inverse_minmax_scale(data_scaled, params):
    """逆变换，恢复原始数据"""
    data_scaled = np.asarray(data_scaled)
    target_min, target_max = params['target_min'], params['target_max']
    # 从 [target_min, target_max] 还原到 [0,1]
    data_01 = (data_scaled - target_min) / (target_max - target_min)
    # 再还原到原始范围
    data_original = data_01 * params['range'] + params['min']
    return data_original

def zscore_scale(data):
    """
    将数据标准化为均值为0，标准差为1。
    返回：
        data_scaled : 标准化后的数据
        params : 包含均值和标准差，用于逆变换
    """
    data = np.asarray(data)
    if data.ndim == 1:
        data = data.reshape(-1, 1)
        single_feature = True
    else:
        single_feature = False

    mean_vals = data.mean(axis=0)
    std_vals = data.std(axis=0)
    std_vals[std_vals == 0] = 1.0  # 避免除零

    data_scaled = (data - mean_vals) / std_vals

    params = {
        'mean': mean_vals,
        'std': std_vals
    }
    if single_feature:
        data_scaled = data_scaled.flatten()
    return data_scaled, params

def inverse_zscore_scale(data_scaled, params):
    """逆变换，恢复原始数据"""
    data_scaled = np.asarray(data_scaled)
    return data_scaled * params['std'] + params['mean']
    
def interpolate_energy(x, y, energy, method = 'cubic', region = 100):
    
    xi = np.linspace(x.min() , x.max(), region)
    yi = np.linspace(y.min() , y.max(), region)
    XI, YI = np.meshgrid(xi, yi)
    
    # 在网格上插值能量值（方法可选 'linear', 'cubic'）
    ZI = griddata((x, y), energy, (XI, YI), method=method )#, fill_value=0)
    return XI, YI, ZI, xi, yi

    
# plot
def plot_3d_landscape(XI, YI, ZI,
                      title = '3D Surface: Energy as f(x, y)',
                      save_file = '3D balabala',
                      cmap = None
                     ):
    fig = plt.figure(figsize=fig_size)
    ax = fig.add_subplot(111, projection='3d')
    surf1 = ax.plot_surface(XI, YI, ZI, cmap=cmap, edgecolor='none', alpha=0.9)
    
    # 添加原始数据点（可选）
    # ax.scatter(x, y, energy, c='black', s=30, depthshade=False)
    # ax.set_zlim(-0.5,2)
    ax.set_xlabel('X')
    ax.set_ylabel('Y')
    ax.set_zlabel('Energy')
    ax.set_axis_off()
    # plt.colorbar(surf1, label='Energy')
    plt.title(title)
    plt.savefig(save_file)
    plt.show()

def plot_3d_landscape_all(XI, YI, 
                      ZI_energy, ZI_rmsd, ZI_solubility,
                      title = '3D Surface: Energy as f(x, y)',
                      save_file = '3D balabala'
                     ):
    fig = plt.figure(figsize=fig_size)
    ax = fig.add_subplot(111, projection='3d')
    surf1 = ax.plot_surface(XI, YI, ZI_energy, cmap='viridis', edgecolor='none', alpha=0.9)
    surf2 = ax.plot_surface(XI, YI, ZI_rmsd, cmap='plasma', edgecolor='none', alpha=0.9)
    surf3 = ax.plot_surface(XI, YI, ZI_solubility, cmap='cividis', edgecolor='none', alpha=0.9)
    
    # 添加原始数据点（可选）
    # ax.scatter(x, y, energy, c='black', s=30, depthshade=False)
    # ax.set_zlim(-0.5,2)
    ax.set_xlabel('X')
    ax.set_ylabel('Y')
    ax.set_zlabel('Energy')
    ax.set_axis_off()
    # plt.colorbar(surf1, label='Energy')
    plt.title(title)
    plt.savefig(save_file)
    plt.show()

def plot_2d_landscape(XI, YI,  ZI,
                      title = '2D Surface: Energy as f(x, y)',
                      save_file = '2D balabala',
                      cmap = None,
                      lable = None,
                      coord = None,
                      E = None,
                      delta = 2
                     ):
    fig = plt.figure(figsize=fig_size)
    contour = plt.contourf(XI, YI, ZI, levels=15, cmap= cmap, alpha = 0.5)
    cbar = plt.colorbar(contour)
    cbar.ax.yaxis.set_major_formatter(FormatStrFormatter('%.0f'))
    if lable is not None:
        print(coord.shape)
        scater_df = pd.DataFrame({'x':coord[:,0], 'y':coord[:,1],'Label':lable})
        sns.scatterplot(data = scater_df, x = 'x', y = 'y', hue = 'Label')
    # plt.xlim([-2,2])
    # plt.ylim([-3,0])

    
    dx = 2
    dy = 2
    
    dE_dy, dE_dx = np.gradient(ZI, dy, dx)
    # plot gradient arrow
    if E == 'U':plt.gca().quiver(XI[::delta, ::delta], YI[::delta, ::delta], dE_dx[::delta, ::delta], dE_dy[::delta, ::delta], color='blue', )
    elif E == 'D':plt.gca().quiver(XI[::delta, ::delta], YI[::delta, ::delta], -dE_dx[::delta, ::delta], -dE_dy[::delta, ::delta], color='blue', )
    else:pass
        
    plt.title(title)
    plt.gca().get_legend().remove()
    plt.gca().set_axis_off()
    plt.savefig(save_file)
    plt.show()

def plot_2d_landscape_arrow(coords, scaled_energy,
                      title = '2D Surface: Energy as f(x, y)',
                      save_file = '2D balabala',
                      cmap = None,
                      lable = None,
                      E = None,
                      delta = 2,
                      start = 19,
                      method = 'cubic',
                      index = 10
                     ):
    # interpolate &  meshgrid

    XI, YI, ZI, x, y = interpolate_energy(coords[:,0], coords[:,1], scaled_energy,     method = method, region = 100)
    # plot mesh
    fig = plt.figure(figsize=fig_size)
    contour = plt.contourf(XI, YI, ZI, levels=15, cmap= cmap, alpha = 0.5)
    cbar = plt.colorbar(contour)
    cbar.ax.yaxis.set_major_formatter(FormatStrFormatter('%.0f'))

    if lable is not None:
        scater_df = pd.DataFrame({'x':coords[:,0], 'y':coords[:,1],'Label':lable})
        sns.scatterplot(data = scater_df, x = 'x', y = 'y', ) #hue = 'Label')
    # plt.gca().get_legend().remove()
    plt.gca().set_axis_off()   
    # gradient
    dx = 0.05
    dy = 0.05
    dE_dy, dE_dx = np.gradient(ZI, dy, dx)
    # interpolate grad
    x0, y0 = coords[index]
    interp_dx = RegularGridInterpolator((y,x), dE_dx)
    interp_dy = RegularGridInterpolator((y,x), dE_dy)
    grad_x0 = interp_dx((y0,x0))   # 注意传入 (y, x) 顺序
    grad_y0 = interp_dy((y0,x0))

    # interp_dx = RegularGridInterpolator((x,y), dE_dx)
    # interp_dy = RegularGridInterpolator((x,y), dE_dy)
    # grad_x0 = interp_dx((x0,y0))   # 注意传入 (y, x) 顺序
    # grad_y0 = interp_dy((x0,y0))
    
    print(grad_x0, grad_y0)
    # target grad
    if E == 'U':descent_vec = np.array([grad_x0, grad_y0])
    elif E == 'D':descent_vec = np.array([-grad_x0, -grad_y0])
    else:return None
        
    # specify the grad arraw
    length = 0.15
    if np.linalg.norm(descent_vec) > 0:
        descent_vec = descent_vec / np.linalg.norm(descent_vec) * length
    # plot the only arrow
    plt.quiver(x0, y0, descent_vec[0], descent_vec[1],
              angles='xy', scale_units='xy', scale=1,
              color='red', width=0.02,)
    plt.title(title)
    plt.savefig(save_file)
    plt.show()
```
