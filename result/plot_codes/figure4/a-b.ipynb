```python
import matplotlib.pyplot as plt
import seaborn
import pandas as pd
import numpy as np
```

```python
# Set the global font size to 7 pt
plt.rcParams.update({
    'font.size': 7,              # Global font size
    'axes.titlesize': 7,          # Title font size
    'axes.labelsize': 7,          # Axis label font size
    'xtick.labelsize': 7,         # X-axis tick font size
    'ytick.labelsize': 7,         # Y-axis tick font size
    'legend.fontsize': 7,         # Legend font size
    'figure.titlesize': 7,        # Figure title font size
})

cm_to_inch = 1 / 2.54
length = width = 7
figsize_cm = (length * cm_to_inch, width * cm_to_inch)  # (9cm, 9cm)
```

```python
csv_file = "data/ss/control-pdn8.csv"
df = pd.read_csv(csv_file)

control_data = df[df.iloc[:, -1] == 'Control']
pdn8_data = df[df.iloc[:, -1] == 'PDN8']

plt.figure(figsize=figsize_cm)

plt.plot(control_data.iloc[:, 0], control_data.iloc[:, 1], label='Control', linewidth=2)
plt.plot(pdn8_data.iloc[:, 0], pdn8_data.iloc[:, 1], label='PDN8', linewidth=2,)

plt.xlabel('Wavelength (nm)')
plt.ylabel('Ellipticity (mdeg)')
ax = plt.gca()

ax.set_xticks(np.arange(200, 255, 10))  # Major tick positions
ax.set_yticks(np.arange(-80, 40, 20))  # Major tick positions

plt.grid(True, linestyle=':', alpha=0.7)
plt.legend()
plt.savefig("control-pdn8.svg",dpi=300, bbox_inches='tight')
plt.show()
```
